```python
# 1 : while문을 사용해 1부터 1000까지의 자연수 중 3의 배수의 합을 구해 보자.
total = 0
i = 1

while (i <= 1000):
    if (i % 3 == 0):
        total = total + i 
    i = i + 1 
print(total)
```

```python
# 6 : 가위바위보 게임에서 승리 횟수가 5회 이상일때 게임이 끝나도록 만들어 보자
# 출력결과 : '승리 : 0/5', 패배 : 1, 비김 : 0'
#'승리 : 1/5', 패배 : 1, 비김 : 0'
#'승리 : 2/5', 패배 : 1, 비김 : 0'
#'승리 : 3/5', 패배 : 1, 비김 : 0'
#'승리 : 3/5', 패배 : 2, 비김 : 0'
#'승리 : 4/5', 패배 : 2, 비김 : 0'
#'승리 : 5/5', 패배 : 2, 비김 : 0'
#'다섯번 이겼습니다. 축하합니다.'
import random

data = ['가위','바위','보']

win = 0      # 승리 횟수
lose = 0     # 패배 횟수
draw = 0   

while win < 5:    
    computer = random.choice(data)

    inputData = input("가위, 바위, 보 중에 입력해주세요 >> ")
    print("플레이어 :", inputData)
    print("컴퓨터 :", computer)
    if inputData not in data:
        print("잘못된 입력입니다. 다시 입력하세요.")
        continue

    
    if (inputData == "가위" and computer == "보") or \
       (inputData == "바위" and computer == "가위") or \
       (inputData == "보" and computer == "바위"):
        win += 1
        print("당신이 이겼습니다.")
    elif inputData == computer:
        draw += 1
        print("비겼습니다.")
    else:
        lose += 1
        print("당신이 졌습니다.")

    # f-string 출력방식 : 문자열 안에 변수를 적기 쉽도록 하기 출력방식
    print(f"'승리 : {win}/5', 패배 : {lose}, 비김 : {draw}'")


print("'다섯번 이겼습니다. 축하합니다.'")
```
